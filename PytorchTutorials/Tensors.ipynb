Tensors are specialized data structures, that are very similar to arrays and matrices. In PyTorch, we use tensors to encode inputs and outputs of a model, as well as the model's parameters.

1. Similar to numpy's ndarrays
2. Can run on GPU's.
3. Tensors and NumPy arrays can often share the same underlying memory, eliminating the need to copy data.
4. Tensors are optimized for automatic differentiation .


```python
import torch
import numpy as np
```

```python
# Creating tensors from data
data  = [[1,2],[3,4]]
x_data = torch.tensor(data)
```

```python
x_data
```

```python
# Creating tensors from numpy arrays
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

```python
x_np
```

```python
# Creating tensors from other tensors
# The new tensor retains the properties (shape, datatype) of of the argument tensor.

x_ones = torch.ones_like(x_data) # retains the properties of x_data
print(f"Ones Tensor: \n {x_ones} \n ")

x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data
print(f"Random Tensor: \n {x_rand} \n")
```

```python
# Shape is a tuple of tensor dimensions. In the functions below, it determines the dimensionality of the output tensor.
shape  = (2,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")


```

```python
#Attributes of a Tensor
print(f"Shape of tensor: {rand_tensor.shape}")
print(f"Datatype of tensor: {rand_tensor.dtype}")
print(f"Device tensor is stored on: {rand_tensor.device}")
```

```python
#Operations on Tensors
# We can perform a variety of operations on tensors.
"""
1. Arithmetic operations
2. Linear algebra
3. Matrix operations
"""
#Each of these operations can be run on CPU or Accelerators like GPU's.
#To run operations on the GPU, we need to use the .to method.
tensor = torch.rand(3,4)
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"
tensor  = tensor.to(device)
x_data = x_data.to(device)
x_ones = x_ones.to(device)

print(f"First row: {tensor[0]} \n")
print(f"First Column: {tensor[:,0]} \n")
print(f"Last Column : {tensor[...,-1]} \n")

print(tensor)
```

```python
# Joining tensors : Sizes of tensors must match except in dimension 0. Expected size 4 but got size 2 for tensor number 1 in the list.

t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

```python
# Arithmetic operations
#Matrix Multiplication :  between 2 Tensors.
# tensor.T : transpose of the tensor.

y1 = tensor @ tensor.T # Matrix Multiplication : Dot Product of 2 Tensors.
y2 = tensor.matmul(tensor.T) 

y1, y2
```

```python
y3 = torch.rand_like(y1)
# print(y3)
torch.matmul(tensor, tensor.T, out = y3)

y3
```

```python
# Element-wise product : Multiplication of 2 Tensors.
z1 = tensor * tensor
z2 = tensor.mul(tensor)

z1, z2
```

```python
z3 = torch.rand_like(tensor)

torch.mul(tensor, tensor, out = z3)
z3

```

```python
# Single element tensors
agg = tensor.sum()
agg_item = agg.item()
print(agg_item, type(agg_item))
```

```python
# In-place operations : Operations that store the result into the operand.
print(tensor, "\n")
tensor.add_(5)
print(tensor)
```

```python
#Bridge with numpy

#Tensors to numpy arrays

t = torch.ones(5)
print(f"t: {t} \n")
n = t.numpy()
print(f"n : {n} \n")

#A change in the tensor reflects in the numpy array and vice versa..
t.add_(1)
print(f"t: {t} \n")
print(f"n : {n} \n")
```
